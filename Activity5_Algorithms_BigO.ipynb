[CELL 3] LINEAR GROWTH

```python
def linear_operations(n):
    count = 0

    for _ in range(n):
        count += 1

    return count


for n in [10, 100, 1000, 5000, 10000]:
    print(f"n={n:5d} | operations={linear_operations(n):5d}")
```

[CELL 4] BIG O CLASSIFICATION

```python
def log_ops(n):
    count = 0

    while n > 1:
        n //= 2
        count += 1

    return count


for n in [10, 100, 1000]:
    print(
        f"n={n:4d} | O(1)=1 | "
        f"O(log n)={log_ops(n):2d} | "
        f"O(n)={n:4d} | "
        f"O(n²)={n*n:7d}"
    )
```

[CELL 7] Linear Search

```python
def linear_search(data, target):
    for i, value in enumerate(data):
        if value == target:
            return i, i + 1

    return -1, len(data)


data = list(range(1, 21))
target = SEED_DIGIT + 1

index, comparisons = linear_search(data, target)

print("Target:", target)
print("Index:", index)
print("Comparisons:", comparisons)
```

[CELL 8} Binary Search

```python
def binary_search(data, target):

    low = 0
    high = len(data) - 1
    comparisons = 0

    while low <= high:

        mid = (low + high) // 2

        comparisons += 1

        if data[mid] == target:
            return mid, comparisons

        elif data[mid] < target:
            low = mid + 1

        else:
            high = mid - 1

    return -1, comparisons


index, comparisons = binary_search(data, target)

print("Target:", target)
print("Index:", index)
print("Comparisons:", comparisons)
```

[CELL 10] Bubble Sort

```python
def bubble_sort(data):

    a = data.copy()
    comparisons = 0
    swaps = 0

    for i in range(len(a)):

        swapped = False

        for j in range(len(a) - i - 1):

            comparisons += 1

            if a[j] > a[j + 1]:

                a[j], a[j + 1] = a[j + 1], a[j]

                swaps += 1
                swapped = True

        if not swapped:
            break

    return a, comparisons, swaps


data = [
    (ID_SUM * 3) % 100,
    (SEED_DIGIT * 11) % 100,
    (NAME_LENGTH * 7) % 100,
    42,
    17,
    89,
    31,
    56
]

sorted_data, comparisons, swaps = bubble_sort(data)

print("Sorted:", sorted_data)
print("Comparisons:", comparisons)
print("Swaps:", swaps)
```

[CELL 11] Selection Sort

```python
def selection_sort(data):

    a = data.copy()
    comparisons = 0
    swaps = 0

    for i in range(len(a) - 1):

        minimum = i

        for j in range(i + 1, len(a)):

            comparisons += 1

            if a[j] < a[minimum]:
                minimum = j

        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            swaps += 1

    return a, comparisons, swaps


sorted_data, comparisons, swaps = selection_sort(data)

print("Sorted:", sorted_data)
print("Comparisons:", comparisons)
print("Swaps:", swaps)
```

[CELL 12] Insertion Sort

```python
def insertion_sort(data):

    a = data.copy()

    comparisons = 0
    shifts = 0

    for i in range(1, len(a)):

        key = a[i]
        j = i - 1

        while j >= 0:

            comparisons += 1

            if a[j] <= key:
                break

            a[j + 1] = a[j]
            shifts += 1

            j -= 1

        a[j + 1] = key

    return a, comparisons, shifts


sorted_data, comparisons, shifts = insertion_sort(data)

print("Sorted:", sorted_data)
print("Comparisons:", comparisons)
print("Shifts:", shifts)
```

[CELL 13] Execution Time Measurement

```python
import time


def measure(sort_function, data, trials=5):

    times = []

    for _ in range(trials):

        start = time.perf_counter()

        result = sort_function(data)

        times.append(
            time.perf_counter() - start
        )

    return result, sum(times) / trials


data_size = 100 + ID_SUM * 10

performance_data = [
    (ID_SUM * 3 + i * 7) % 1000
    for i in range(data_size)
]

bubble_result, bubble_time = measure(
    bubble_sort,
    performance_data
)

selection_result, selection_time = measure(
    selection_sort,
    performance_data
)

insertion_result, insertion_time = measure(
    insertion_sort,
    performance_data
)

print(f"Dataset Size: {data_size}")

print(
    f"Bubble Sort: "
    f"{bubble_time:.8f} s"
)

print(
    f"Selection Sort: "
    f"{selection_time:.8f} s"
)

print(
    f"Insertion Sort: "
    f"{insertion_time:.8f} s"
)
```
